# Fashion Intelligence: Problem Definition

This notebook defines the problem before any data result or modelling choice is known. It fixes the users, prediction units, required outputs, evaluation boundary, and ownership rules that later notebooks must follow. It contains no code, EDA finding, model result, or selected metric.

## 1. Executive summary

The project turns fashion images into four catalogue predictions and a visual-search result. A useful submission must do more than produce scores: it must compare approaches, show where they fail, protect independent evaluation, and explain which method is suitable for each task.

Notebook 01 will discover the actual data facts. Task notebooks will choose their own preprocessing, metrics, models, and experiment scope after that evidence is available.

## 2. Real-world users and decisions supported

Potential users include catalogue teams, search users, and engineers who operate the final system. The system supports three kinds of decision:

- add or check catalogue attributes from a product image;
- find visually related products for a query image;
- decide whether a model is reliable, efficient, and safe enough for its intended use.

The model output is decision support. Ambiguous labels and visually similar products still need honest uncertainty and failure analysis.

## 3. Problem statement and prediction unit

Given a fashion image, learn image-based systems for the required classification and visual search tasks. The main reporting unit is one product ID. Multiple files or transformed views of one product must never be counted as independent products.

For classification, one product image maps to one output for the selected target. For visual search, one query image maps to an ordered list of product IDs.

## 4. Assignment tasks

The four tasks share the protected data contract but own their task-specific decisions.

### 4.1 Task 1 - Article type classification

Predict the `articleType` value for each product image. The Task 1 owner must later define the preprocessing comparison, metric rule, baselines, model candidates, and rare-class analysis.

### 4.2 Task 2 - Season classification

Predict the `season` value for each product image. The Task 2 owner must later test whether the image carries enough season signal and examine shortcut, ambiguity, and calibration risk.

### 4.3 Task 3 - Gender and usage classification

Predict `gender` and `usage` as two separate required outputs. The Task 3 owner may compare separate and shared representations, but must keep missing-label masks and report negative transfer if a shared design harms either output.

### 4.4 Task 4 - Fashion visual search

Given a query fashion image, return an ordered Top-K list of visually similar product IDs. The Task 4 owner must later define query handling, gallery construction, image-size strategy, relevance evidence, ranking rule, cutoff, and latency checks before comparing approaches.

## 5. Inputs available at prediction time

The required model input is image data. A product ID may be used to join files and write an output, but target labels and protected outcomes are not prediction-time features. Any extra metadata or external image source must be treated as a separately justified task decision, not an invisible shared default.

## 6. Required outputs

Classification must produce the four columns required by the fixed prediction CSV: `gender`, `articleType`, `season`, and `usage`, joined to `id`. Task 4 produces a ranked list of catalogue product IDs for a query image and is evaluated through its code, evidence, application, and report rather than extra prediction-CSV columns.

## 7. Evaluation principles and independent evidence

Development data may support EDA, training, validation, and method choice. A separate internal holdout remains sealed until every task has frozen its method. Duplicate and family groups must not cross an evaluation boundary. Official prediction data is a separate unlabeled role.

A task owner must choose and document a primary metric before comparing results. The choice must match the task and be supported by secondary evidence, error analysis, robustness, and cost. No exact metric is selected in this notebook.

## 8. Success dimensions and ownership

A strong result explains:

- predictive or ranking quality under one frozen comparison protocol;
- performance across common, rare, ambiguous, and failure slices;
- stability across the declared validation evidence;
- model size, inference cost, and practical limits;
- why the selected method is preferable to the compared alternatives.

Notebook 01 owns shared data facts. Notebooks 02-05 own task choices and experiments. Notebook 06 owns final refit, the one-time holdout evaluation, and ultimate judgement.

## 9. Constraints

- Submitted models are trained from scratch; pretrained systems are comparison benchmarks only.
- `data/processed/splits.csv` is the only shared split.
- Training runs must use the shared run registry.
- The prediction CSV schema and target names stay unchanged.
- Protected labels cannot guide EDA, preprocessing, tuning, or model selection.
- Claims must be reproducible from code, artifacts, and recorded run IDs.

## 10. Assumptions, risks, and failure costs

Catalogue labels may describe business categories that are not fully visible in pixels. Some classes may be rare, overlapping, missing, or visually ambiguous. Duplicate products can make evaluation look better than real use. Query images may differ in size or source from training images.

Wrong predictions can create bad catalogue fields or irrelevant search results. The final judgement must therefore name affected users, visible failure modes, and cases that need human review rather than hiding them behind one aggregate score.

## 11. Non-goals

This stage does not clean data, select an image transformation, handle imbalance, choose a model, choose a metric, train anything, build a search index, inspect holdout outcomes, or claim that any task is solved.

## 12. Deliverables and notebook handoff

The workflow is:

1. define the problem here;
2. audit and prepare the shared teacher data in Notebook 01;
3. make and record task-owned choices in Notebooks 02-05;
4. register every experiment and freeze the selected method;
5. refit on the permitted development data;
6. perform one independent evaluation in Notebook 06;
7. create the fixed-format predictions, application handoff, and report evidence.

## 13. Problem-definition readiness gate

Proceed to Data Preparation only when the team can answer all of these without using an EDA result:

- Who uses each output and for what decision?
- What is one prediction or retrieval unit?
- Which inputs exist at prediction time?
- Which outputs are required?
- Which data role is protected?
- Who owns preprocessing, metric, model, and final-evaluation choices?

All six answers are defined above. Notebook 01 may now measure the actual dataset and create the shared handoff artifacts.
